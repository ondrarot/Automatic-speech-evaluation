```python
import utmosv2
import pandas as pd
import os
import torch

# 1. Verify the GPU is working!
print("Is CUDA (GPU) available?:", torch.cuda.is_available())

# 2. Setup paths on the MetaCentrum storage
data_dir = "/data_dir"
output_csv = "/results.csv"

def main():
    print("Downloading/Loading UTMOSv2 model on GPU...")
    # We explicitly tell it to use the GPU ("cuda")
    model = utmosv2.create_model(pretrained=True, device="cuda")
    
    print(f"Evaluating the wav files in: {data_dir}")

    results = model.predict(input_dir=data_dir, device="cuda")
    
    # 3. Clean up the CSV formatting
    df = pd.DataFrame(results)
    df['file_path'] = df['file_path'].apply(os.path.basename)
    df = df.rename(columns={'file_path': 'file_name'})
    
    # Save the results
    df.to_csv(output_csv, index=False)
    
    print(f"Success! Evaluated {len(df)} files.")
    print(f"Results saved to {output_csv}")

if __name__ == "__main__":
    main()
```
